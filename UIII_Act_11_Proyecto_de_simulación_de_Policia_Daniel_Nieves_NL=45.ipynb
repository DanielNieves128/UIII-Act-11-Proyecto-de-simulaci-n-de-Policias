```python
# ============================================
# SISTEMA INTELIGENTE DE POLICÍAS
# Daniel Nieves NL=45
# Profesor: Eliseo Nava
# Materia: Elabora Proyectos Con Programacion Logica
# Cbtis128
# Especialidad: Inteligencia Artificial (IA)
# ============================================
# Lista de diccionarios (inicia vacía)
policias = []
# ============================================
# FUNCION PARA AGREGAR POLICÍAS
# ============================================
def agregar_policia():
    global policias # Declare that we intend to modify the global 'policias' list
    nuevo_policias = {} # Create a local dictionary for the new police data
    print("\n===== REGISTRO DE POLICÍA ====")
    nuevo_policias["numero_placa"] = input("Número de placa: ")
    nuevo_policias["patrulla_asignada"] = input("Patrulla asignada: ")
    nuevo_policias["sector_vigilancia"] = input("Sector de vigilancia: ")
    nuevo_policias["anios_servicio"] = input("Años de servicio: ")
    nuevo_policias["rango"] = input("Rango: ")
    nuevo_policias["curso_especializacion"] = input("Curso de especialización: ")
    policias.append(nuevo_policias) # Append the new police dictionary to the global list
    print("\nPolicía agregado correctamente.\n")
# ============================================
# FUNCION PARA MOSTRAR POLICÍAS
# ============================================
def mostrar_policias():
    print("\n============= LISTADO DE POLICÍAS ===========telek=")
    if len(policias) == 0:
        print("No hay policías registrados.\n")
    else:
        print("-" * 125)
        print(
            f"{'PLACA':<15}"
            f"{'PATRULLA':<15}"
            f"{'SECTOR':<20}"
            f"{'AÑOS':<10}"
            f"{'RANGO':<20}"
            f"{'CURSO':<35}"
        )
        print("-" * 125)
        for p in policias:
            print(
                f"{p['numero_placa']:<15}"
                f"{p['patrulla_asignada']:<15}"
                f"{p['sector_vigilancia']:<20}"
                f"{p['anios_servicio']:<10}"
                f"{p['rango']:<20}"
                f"{p['curso_especializacion']:<35}"
            )
        print("-" * 125)
# ============================================
# FUNCION PARA BUSCAR POLICÍA
# ============================================
def buscar_policia():
    buscar = input("\nIngrese el número de placa a buscar: ")
    encontrado = False
    for p in policias:
        if p["numero_placa"] == buscar:
            print("\nPOLICÍA ENCONTRADO")
            print("---------------------------")
            print("Número de placa:", p["numero_placa"])
            print("Patrulla asignada:", p["patrulla_asignada"])
            print("Sector de vigilancia:", p["sector_vigilancia"])
            print("Años de servicio:", p["anios_servicio"])
            print("Rango:", p["rango"])
            print("Curso:", p["curso_especializacion"])
            encontrado = True
    if encontrado == False:
        print("\nNo se encontró un policía con esa placa.")
# ============================================
# FUNCION PARA ELIMINAR POLICÍA
# ============================================
def eliminar_policia():
    eliminar = input("\nIngrese el número de placa a eliminar: ")
    encontrado = False
    # Iterate through a copy of the list when modifying it
    for p in list(policias):
        if p["numero_placa"] == eliminar:
            policias.remove(p)
            print("\nPolicía eliminado correctamente.")
            encontrado = True
            break # Exit loop after finding and removing the police
    if not encontrado:
        print("\nNo existe un policía con esa placa.")
# ============================================
# FUNCION DE ANALISIS INTELIGENTE
# ============================================
def analisis_inteligente():
    print("\n===== ANALISIS INTELIGENTE ===telek=")
    if len(policias) == 0:
        print("No hay policías registrados.")
    else:
        for p in policias:
            print("\nPolicía:", p["numero_placa"])
            if int(p["anios_servicio"]) >= 10:
                print("Policía con mucha experiencia.")
            elif int(p["anios_servicio"]) >= 5:
                print("Policía con experiencia media.")
            else:
                print("Policía nuevo en servicio.")
            if p["rango"].lower() == "comandante":
                print("Nivel alto de autoridad.")
            elif p["rango"].lower() == "oficial":
                print("Oficial operativo.")
            else:
                print("Rango registrado:", p["rango"])
# ============================================
# MENU PRINCIPAL
# ============================================
while True:
    print("\n===================================")
    print("     SISTEMA INTELIGENTE POLICIAL")
    print("===================================")
    print("1. Agregar policía")
    print("2. Mostrar policías")
    print("3. Buscar policía")
    print("4. Eliminar policía")
    print("5. Análisis inteligente")
    print("6. Salir")
    opcion = input("\nSeleccione una opción: ")
    # ========================================
    # CONDICIONES
    # ========================================
    if opcion == "1":
        agregar_policia()
    elif opcion == "2":
        mostrar_policias()
    elif opcion == "3":
        buscar_policia()
    elif opcion == "4":
        eliminar_policia()
    elif opcion == "5":
        analisis_inteligente()
    elif opcion == "6":
        print("\nGracias por usar el sistema.")
        print("Programa creado por Daniel Nieves NL=45")
        break

    else:
        print("\nOpción no válida.")
```
